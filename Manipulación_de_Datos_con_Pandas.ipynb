## Introducción al Análisis de Datos

### ¿Qué es Pandas?

Pandas es una biblioteca de Python especializada en la manipulación, transformación y análisis de datos.

Mientras que NumPy está orientado al trabajo eficiente con datos numéricos en arrays, Pandas incorpora estructuras diseñadas para trabajar con datos similares a los que encontramos en un dataset real. Por ejemplo:

- pasajeros de un barco
- transacciones bancarias
- ventas de una tienda
- resultados de encuestas
- registros médicos
- datos meteorológicos

Una tabla puede contener:

- **filas** → registros u observaciones
- **columnas** → variables o características
- **valores** → información correspondiente a cada registro

La estructura principal de Pandas es el **DataFrame**: una tabla de datos que podemos manipular mediante Python.

### ¿Por qué Pandas es importante en Análisis de Datos?

Pandas permite realizar muchas de las tareas que aparecen habitualmente durante un análisis:

1. Cargar datos
2. Explorar un dataset
3. Comprender sus variables
4. Detectar valores faltantes
5. Seleccionar información
6. Filtrar registros
7. Crear nuevas variables
8. Transformar datos
9. Agrupar información
10. Calcular estadísticas
11. Ordenar resultados
12. Combinar datasets
13. Preparar los datos para visualizaciones
14. Obtener información que permita responder preguntas

Por eso Pandas ocupa un lugar central dentro del proceso de análisis de datos.

### Pandas y NumPy

Pandas está estrechamente relacionado con NumPy.

- **NumPy** → operaciones numéricas eficientes
- **Pandas** → manipulación y análisis de datos tabulares

NumPy trabaja muy bien con arrays como `[22, 38, 26, 35, 28]`, mientras que Pandas permite trabajar con información tabular como:

| PassengerId | Survived | Pclass | Name | Sex | Age | Fare | Embarked |
|---|---|---|---|---|---|---|---|
| 1 | 0 | 3 | Braund | male | 22 | 7.25 | S |
| 2 | 1 | 1 | Cumings | female | 38 | 71.28 | C |
| 3 | 1 | 3 | Heikkinen | female | 26 | 7.92 | S |

Esto se aproxima mucho más a los datasets que encontraremos en un análisis real.

Una de las tareas más importantes del análisis de datos es seleccionar solamente los registros que cumplen determinadas condiciones. Por ejemplo, pasajeros mayores de 40 años.

> **¿Para qué le sirve al analista?** Permite concentrar el análisis en un subconjunto específico de los datos.

Podemos guardar la condición en una variable, lo que hace explícito el concepto de **máscara booleana**.

> **¿Para qué le sirve al analista?** Permite analizar exclusivamente un grupo determinado.
>
> **Pregunta:** «¿Cómo eran las pasajeras mujeres?»

> **¿Para qué le sirve al analista?** Permite localizar pasajeros que pagaron tarifas elevadas.
>
> **Pregunta:** «¿Qué pasajeros pagaron más de 100 dólares?»

### Filtrar utilizando varias condiciones (AND)

Podemos combinar condiciones utilizando `&`. Esto significa: «Edad mayor a 40 Y tarifa superior a 100.»

> **¿Para qué le sirve al analista?** Permite construir consultas más específicas.
>
> **Pregunta:** «¿Qué pasajeros mayores de 40 años pagaron más de 100 dólares?»

> **¿Para qué le sirve al analista?** Permite analizar grupos formados por diferentes categorías.

````python
df[
    (df["Embarked"] == "C") |
    (df["Embarked"] == "Q")
]
````

### `isin()`

Cuando queremos seleccionar varias categorías.

> **¿Para qué le sirve al analista?** Hace que las consultas sobre múltiples categorías sean más claras y legibles.

````python
df[
    df["Embarked"].isin(["C", "Q", "S"])
]
````

### Filtrar por rango

> **¿Para qué le sirve al analista?** Permite estudiar segmentos específicos de una población.
>
> **Pregunta:** «¿Cómo eran los pasajeros de entre 20 y 30 años?»

````python
df[
    (df["Age"] >= 20) &
    (df["Age"] <= 30)
]
````

## Operaciones vectorizadas

Una de las ventajas de Pandas es que podemos aplicar operaciones a una columna completa, sin necesidad de recorrer cada fila mediante un `for`.

> **¿Para qué le sirve al analista?** Permite transformar grandes cantidades de datos de manera compacta y eficiente.

````python
df["Fare"] * 1.1
````

````python
# Operación vectorizada con un escalar (factor de inflación)

# Pandas toma el escalar 130 y lo multiplica por cada uno de los 891 precios
# de la columna Fare simultáneamente. Un pasajero que pagó £7 en 1912, ahora
# verá £910 en la nueva columna.
df["Fare_2025"] = df["Fare"] * 130
df
````

````python
# Operación vectorizada: restar la media de toda la columna a cada valor

# Pandas calcula la media de toda la columna Age (un solo número) y lo resta
# de cada uno de los 891 valores de edad al mismo tiempo. Obtienes una columna
# nueva con valores como -5.2 (más joven que el promedio) o +8.7 (más viejo).

df["Edad_diferencia"] = df["Age"] - df["Age"].mean()
df
````

## Crear categorías

Supongamos que queremos clasificar a los pasajeros por rango de edad.

> **¿Para qué le sirve al analista?** Permite transformar valores numéricos en categorías útiles para el análisis (niño, adulto, senior), lo que facilita posteriormente la segmentación.

````python
df["AgeGroup"] = "Adulto"

df.loc[
    df["Age"] < 18,
    "AgeGroup"
] = "Niño"

df.loc[
    df["Age"] >= 60,
    "AgeGroup"
] = "Anciano"
````

## `groupby()`

Una de las funciones más importantes de Pandas: permite dividir los datos en grupos según una variable.

```
Todos los pasajeros
        ↓
-------------------
Hombre
Mujer
```

> **¿Para qué le sirve al analista?** Permite comparar el comportamiento de una métrica entre diferentes grupos. Esta es una de las operaciones más importantes del análisis de datos.
>
> **Pregunta:** «¿Cómo se comporta la supervivencia según el sexo?»

````python
df.groupby("Sex")
````

### `groupby()` + `mean()`

> **¿Para qué le sirve al analista?** Permite comparar la tasa de supervivencia entre grupos.
>
> **Pregunta de negocio:** «¿Las mujeres tenían más probabilidades de sobrevivir que los hombres?»

````python
df.groupby("Sex")["Survived"].mean()
````

### `groupby()` + `sum()`

> **¿Para qué le sirve al analista?** Permite conocer cuántos sobrevivieron en cada grupo.
>
> **Pregunta:** «¿Cuántas mujeres y cuántos hombres sobrevivieron?»

````python
df.groupby("Sex")["Survived"].sum()
````

### `groupby()` + `count()`

> **¿Para qué le sirve al analista?** Permite conocer la cantidad de personas en cada categoría.

````python
df.groupby("Sex")["PassengerId"].count()
````

### `groupby()` + múltiples estadísticas

> **¿Para qué le sirve al analista?** Permite construir un perfil estadístico de cada categoría, y comparar grupos desde diferentes perspectivas.

````python
df.groupby("Sex")["Fare"].agg(
    ["count", "mean", "min", "max", "sum"]
)
````

````python
# Ejemplo más flexible en donde obtenemos las estadísticas distintas
# para Fare y para Age en la misma tabla.

df.groupby("Sex").agg({
    "Fare": ["mean", "max"],
    "Age": ["mean", "min"]
})
````

### Agrupaciones múltiples

También podemos agrupar utilizando más de una variable.

> **¿Para qué le sirve al analista?** Permite detectar diferencias más específicas.
>
> **Pregunta:** «¿Cuál es la tasa de supervivencia según sexo y clase?» Este tipo de análisis comienza a acercarnos a preguntas más interesantes.

````python
df.groupby(
    ["Sex", "Pclass"]
)["Survived"].mean()
````

Genera una tabla de frecuencias cruzadas.

> **¿Para qué le sirve al analista?** Permite comparar dos variables categóricas.

> **Pregunta:** «¿Cómo se distribuyen los sobrevivientes según el sexo?»

## Tablas dinámicas (pivot tables)

> **¿Para qué le sirve al analista?** Una tabla dinámica permite resumir información de manera muy flexible. Es conceptualmente similar a una tabla dinámica de Excel, pero construida mediante código.

## Un análisis un poco más completo

Podemos combinar varias operaciones. Por ejemplo, un resumen de supervivencia por clase y sexo, ordenado de mayor a menor tasa:

1. `groupby()` → agrupar por clase y sexo
2. `agg()` → calcular cantidad y tasa
3. `sort_values()` → ordenar los resultados

Esto representa mucho mejor cómo se utiliza Pandas en un análisis real.

````python
resultado = (
    df.groupby(["clase", "sexo"])
      ["sobrevivido"]
      .agg(["count", "mean"])
      .sort_values("mean", ascending=False)
)
resultado
````
