#Pandas

- Pandas (PANel DAta) es una librería de Python que proporciona herramientas de análisis y manipulación de datos de alto rendimiento utilizando sus potentes estructuras de datos.
- El nombre de Pandas se deriva del término “Panel Data” y es la librería por excelencia de análisis de datos de Python ya que con ella podemos crear distintos dataframes.
- Un DataFrame es una estructura de datos con dos dimensiones en la cual se puede guardar datos de distintos tipos (como caracteres, enteros, valores de punto flotante y más) en columnas.
- Si se necesita una biblioteca de Python enfocada en los datos científicos, Pandas es una de las más útiles.
- Página oficial: https://pandas.pydata.org/



- Es una librería muy sencilla de usar para manipular datos y se utiliza en campos como finanzas, ciencias sociales, estadística e ingeniería.
- Sus principales funciones se conforman por **Series de datos** y el **Data Frame** para dos dimensiones.
- Define nuevas estructuras de datos basadas en los arrays de la librería NumPy pero con nuevas funcionalidades.
- Permite leer y escribir fácilmente ficheros en formato CSV, Excel y bases de datos SQL y permite acceder a los datos mediante índices o nombres para filas y columnas.
- Ofrece métodos para reordenar, dividir y combinar conjuntos de datos.
- Permite trabajar con series temporales.
- Realiza todas estas operaciones de manera muy eficiente.

#Tipos de datos en pandas

- Pandas proporciona dos tipos de clases para manejar datos:
1. Series: una matriz unidimensional que contiene datos de cualquier tipo como enteros, cadenas, objetos Python, etc.

2. DataFrame: una estructura de datos bidimensional que contiene datos como una matriz bidimensional o una tabla con filas y columnas.




##Creación de una serie

```python
import numpy as np
import pandas as pd
```

```python
# Creación de series usando listas
# El índice por defecto se crea con números enteros positivos empezando desde 0.
serie1 = pd.Series([1, 2, 3])
print(serie1)

# Se pueden modificar los indices.
# la cantidad de indices debe ser igual a la cantidad de valores de la serie.
serie1 = pd.Series([1, 2, 3], index=["a", "b", "c"])
print(serie1)

# esto dara un error
# serie1 = pd.Series([1, 2, 3], index=["a", "b"])
# print(serie1)
```

```python
# Creación de series usando range
serie2 = pd.Series(range(1, 4))

print(serie2)
```

```python
# Creación de series usando un diccionario con etiquetas y valores
items = {'a': 1, 'b': 2, 'c': 3}
serie3 = pd.Series(items)

print(serie3)
```

```python
# Todas las series que hemos visto hasta ahora no tienen asignado ningún nombre.
# Lo podemos hacer usando el parámetro name en la creación de la serie
serie4 = pd.Series([1, 2, 3], name='Numeros enteros')

print(serie4)
```

###Atributos de una serie

```python
# Las series en pandas contienen gran cantidad de atributos. A continuación destacaremos algunos de ellos
data = {'Apple': 147000,
        'Samsung': 267937,
        'Google': 135301,
        'Microsoft': 163000,
        'Huawei': 197000,
        'Dell': 158000,
        'Facebook': 58604,
        'Foxconn': 878429,
        'Sony': 109700
    }

empleados = pd.Series(data, name='Cantidad empleados Tech')

print(empleados)
print("-" * 40)

# para ver los indices de la Serie:
print(empleados.index)
print("-" * 40)

# para ver los valores de la Serie:
print(empleados.values)
print("-" * 40)

# para ver el tipo de dato de la Serie:
print(empleados.dtype)
print("-" * 40)

# para ver el nombre de la Serie:
print(empleados.name)
print("-" * 40)

# para ver la memoria ocupada de la Serie:
print(empleados.nbytes)
print("-" * 40)

# para ver el numero de registros de la Serie:
print(empleados.size)
print("-" * 40)

# para cambiar el nombre de la Serie:
s = pd.Series([44, 46, 23, 76], name="Serie de numeros")
s = s.rename("Cambio el nombre")
print(s)
print("-" * 40)

# cambiar el nombre de los indices de una Serie.
# Si queremos podemos cambiar algunos nombres nada mas, como en este ejemplo.
s1 = pd.Series([111, 543, 878, 99, 54])
s1 = s1.rename(index={0: 'x', 1: 'y', 2: 'z'})
print(s1)
```

```python
# drop() para eliminar elementos por etiqueta de índice o por posición.
data = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
serie = pd.Series(data, name='ejemplo')

print("Serie original:")
print(serie)

serie_modificada = serie.drop('b')
print("\nSerie después de eliminar 'b':")
print(serie_modificada)

# aca eliminamos el indice 0
serie_modificada_dos = serie.drop(serie.index[0])
print(serie_modificada_dos)
```

###Seleccion usando indexado numerico

```python
# Para acceder a los registros por su posición (índice numérico) basta usar corchetes como ya se ha visto en cualquier secuencia
data = {'Apple': 147000,
        'Samsung': 267937,
        'Google': 135301,
        'Microsoft': 163000,
        'Huawei': 197000,
        'Dell': 158000,
        'Facebook': 58604,
        'Foxconn': 878429,
        'Sony': 109700}

empleados = pd.Series(data, name='Cantidad empleados Tech')

print(empleados[0])
print("-" * 40)
print(empleados[-1])
print("-" * 40)
print(empleados[2:5])
print("-" * 40)
print(empleados[1:6:2])

print("-" * 40)
# de esta forma podemos acceder y mostrar tanto el indice como el valor.
print(empleados[empleados.index == 'Dell'])
```

```python
# El atributo iloc es un alias (algo más expresivo) que permite realizar las
# mismas operaciones de indexado (con corchetes) que hemos visto anteriormente.
# La palabra "iloc" es una abreviatura de "integer location" (ubicación entera).
data = {'Apple': 147000,
        'Samsung': 267937,
        'Google': 135301,
        'Microsoft': 163000,
        'Huawei': 197000,
        'Dell': 158000,
        'Facebook': 58604,
        'Foxconn': 878429,
        'Sony': 109700}

empleados = pd.Series(data, name='Cantidad empleados Tech')

print(empleados.iloc[6])
print("-" * 40)
print(empleados.iloc[1:4])
print("-" * 40)
print(empleados.iloc[1:6:2])
```

###Selección usando etiquetas

```python
# En el caso de aquellas series que dispongan de un índice con etiquetas,
# podemos acceder a sus registros utilizando las mismas

data = {'Apple': 147000,
        'Samsung': 267937,
        'Google': 135301,
        'Microsoft': 163000,
        'Huawei': 197000,
        'Dell': 158000,
        'Facebook': 58604,
        'Foxconn': 878429,
        'Sony': 109700}

empleados = pd.Series(data, name='Cantidad empleados Tech')

print(empleados["Google"])
print("-" * 40)
print(empleados["Microsoft":"Foxconn"])
print("-" * 40)
print(empleados["Samsung":"Sony":2])
```

```python
# El atributo loc es un alias (algo más expresivo) que permite realizar las
# mismas operaciones de indexado (con corchetes) que hemos visto anteriormente
# La palabra "loc" es una abreviatura de "location" (ubicación).
data = {'Apple': 147000,
        'Samsung': 267937,
        'Google': 135301,
        'Microsoft': 163000,
        'Huawei': 197000,
        'Dell': 158000,
        'Facebook': 58604,
        'Foxconn': 878429,
        'Sony': 109700}

empleados = pd.Series(data, name='Cantidad empleados Tech')

print(empleados.loc["Apple"])
print("-" * 40)
print(empleados.loc["Microsoft":"Foxconn"])
print("-" * 40)
print(empleados.loc["Samsung":"Sony":2])
```

###Fragmentos de comienzo y fin

```python
# A nivel exploratorio, es bastante cómodo acceder a una porción inicial (o final)
# de los datos que manejamos. Esto se puede hacer de forma muy sencilla con series
data = {'Apple': 147000,
        'Samsung': 267937,
        'Google': 135301,
        'Microsoft': 163000,
        'Huawei': 197000,
        'Dell': 158000,
        'Facebook': 58604,
        'Foxconn': 878429,
        'Sony': 109700}

empleados = pd.Series(data, name='Cantidad empleados Tech')

# head() es un método que se utiliza para mostrar las primeras filas de un
# DataFrame o las primeras observaciones de una Serie.
print(empleados.head(3))

print("-" * 40)

# tail() se utiliza para mostrar las últimas filas de un DataFrame o las últimas
# observaciones de una Serie.
print(empleados.tail(2))
```

###Operaciones con series
####Si tenemos en cuenta que una serie contiene valores en formato ndarray podemos concluir que las operaciones sobre arrays (como lo haciamos con Numpy) son aplicables al caso de las series.

####OPERACIONES LÓGICAS

```python
# Supongamos que queremos filtrar aquellas empresas que tengan más de 200000 trabajadores/as
data = {'Apple': 147000,
        'Samsung': 267937,
        'Google': 135301,
        'Microsoft': 163000,
        'Huawei': 197000,
        'Dell': 158000,
        'Facebook': 58604,
        'Foxconn': 878429,
        'Sony': 109700}

empleados = pd.Series(data, name='Cantidad empleados Tech')

print(empleados > 200_000)

print("-" * 40)

# tambien podemos usar operadores logicos
print((empleados > 200_000) & (empleados < 400_000))

print("-" * 40)

# Hemos obtenido una serie «booleana». Si queremos aplicar esta «máscara»,
# podemos hacerlo con indexado
print(empleados[empleados > 200_000])
```

####ORDENACIÓN

```python
# Ordenación de una serie por sus valores
# sort_values() se utiliza para ordenar los valores de un DataFrame o una Serie
# según un criterio específico.
data = {'Apple': 147000,
        'Samsung': 267937,
        'Google': 135301,
        'Microsoft': 163000,
        'Huawei': 197000,
        'Dell': 158000,
        'Facebook': 58604,
        'Foxconn': 878429,
        'Sony': 109700}

empleados = pd.Series(data, name='Cantidad empleados Tech')

print(empleados.sort_values())

print("-" * 40)

# tambien lo podemos hacer con un orden descendente
print(empleados.sort_values(ascending=False))

# el parametro inplace se utiliza para especificar si deseas realizar la operación
# de ordenamiento directamente en el DataFrame o la Serie existente o si prefieres
# crear un nuevo DataFrame o Serie con los valores ordenados sin modificar el original.
# Cuando inplace=True, los métodos ordenarán el DataFrame o la Serie directamente
# y no devolverán un nuevo objeto. En este caso, el DataFrame o Serie original
# se modificará y no se devolverá nada (o se devolverá None en la práctica).
empleados.sort_values(inplace=True)

print("-" * 40)

print(empleados)
```

```python
# sort_index() se utiliza para ordenar los índices de un DataFrame o una Serie.
# El índice es la etiqueta que identifica cada fila en un DataFrame o cada elemento en una Serie.
# sort_index() permite organizar estos índices en un orden específico, ya sea
# ascendente o descendente.
data = {'Apple': 147000,
        'Samsung': 267937,
        'Google': 135301,
        'Microsoft': 163000,
        'Huawei': 197000,
        'Dell': 158000,
        'Facebook': 58604,
        'Foxconn': 878429,
        'Sony': 109700}

empleados = pd.Series(data, name='Cantidad empleados Tech')

# Ordenación de una serie por sus indices
print(empleados.sort_index())

print("-" * 40)

# tambien lo podemos hacer con un orden descendente
print(empleados.sort_index(ascending=False))
```

####CONTANDO VALORES

```python
# Si queremos obtener una «tabla de frecuencias» podemos contar los valores que
# existen en nuestra serie
numeros = pd.Series([5, 5, 3, 6, 5, 2, 8, 3, 8, 7, 6])

print(numeros.value_counts())

# value_counts() se utiliza para contar las ocurrencias de valores en una serie de datos.
```

```python
# Vinculado con el caso anterior, podemos obtener el número de valores únicos en la serie
print(numeros.nunique())
```

```python
# count() devuelve el número de valores «no nulos» que contiene la serie:
numeros = pd.Series([5, 5, 3, 6, 5, 2, 8, 3, 8, 7, 6])

print(numeros.count())

# probamos con algunos Null
numeros2 = pd.Series([5, None, np.nan, None, 5, 2, np.nan, 3, 8, 7, 6])
print(numeros2.count())
```

####OPERACIONES ARITMÉTICAS

```python
# OPERACIONES ENTRE SERIES Y ESCALARES
# Podemos operar entre series y escalares sin ningún tipo de problema
data = {'Apple': 147000,
        'Samsung': 267937,
        'Google': 135301,
        'Microsoft': 163000,
        'Huawei': 197000,
        'Dell': 158000,
        'Facebook': 58604,
        'Foxconn': 878429,
        'Sony': 109700}

empleados = pd.Series(data, name='Cantidad empleados Tech')

print(empleados / 100)
print("-" * 40)
print(empleados * 100)
print("-" * 40)
print(empleados + 500)
print("-" * 40)
print(empleados["Apple"] + 500) # sumo 500 empleados a Apple
```

```python
# Para el caso de operaciones entre series, vamos a ejemplificarlo con las dos siguientes
# Tener en cuenta que las operaciones se realizan entre registros que tienen el mismo índice (etiqueta).
# En este ejemplo queremos calcular la relacion ingreso por empleado en cada empresa.
data = {'Apple': 147000,
        'Samsung': 267937,
        'Google': 135301,
        'Microsoft': 163000,
        'Huawei': 197000,
        'Dell': 158000,
        'Facebook': 58604,
        'Foxconn': 878429,
        'Sony': 109700}

data1 = {'Apple': 274515,
        'Samsung': 200734,
        'Google': 182527,
        'Microsoft': 143015,
        'Huawei': 129184,
        'Dell': 92224,
        'Facebook': 85965,
        'Foxconn': 181945,
        'Sony': 84893}

empleados = pd.Series(data, name='Cantidad empleados Tech')
ingresos = pd.Series(data1, name='Ingresos empleados Tech')

print(ingresos/empleados)
```

```python
# otro ejemplo
numeros1 = pd.Series([1, 2, 3, 4, 5])
numeros2 = pd.Series([8, 9, 10, 11])

print(numeros1 * numeros2)

# Cuando multiplicas dos Series de Pandas de igual longitud, la multiplicación
# se realiza elemento por elemento en función de la posición. En este caso, la
# operación numeros1 * numeros2 multiplica el primer elemento de numeros1 por el
# primer elemento de numeros2, el segundo por el segundo, y así sucesivamente.

# Sin embargo, hay una diferencia en la longitud de las dos series.
# La serie numeros1 tiene 5 elementos, mientras que numeros2 tiene 4 elementos.
# Cuando se realiza una operación de multiplicación entre dos series de longitudes
# diferentes, Pandas trata de alinearlas alineando los índices y rellenando con
# valores NaN (Not a Number) en las posiciones donde no hay coincidencia de índices.
```

####FUNCIONES ESTADÍSTICAS

```python
# Existen multitud de funciones estadísticas que podemos aplicar a una serie.
# Dependiendo del tipo de dato con el que estamos trabajando, serán más útiles
# unas que otras. Veamos dos funciones a modo de ejemplo

data = {'Apple': 147000,
        'Samsung': 267937,
        'Google': 135301,
        'Microsoft': 163000,
        'Huawei': 197000,
        'Dell': 158000,
        'Facebook': 58604,
        'Foxconn': 878429,
        'Sony': 109700}

empleados = pd.Series(data, name='Cantidad empleados Tech')

# mean(): Calcula la media de los valores.
print(empleados.mean())
print("-" * 40)

# median(): Calcula la mediana de los valores.
print(empleados.median())
print("-" * 40)

# std() / var(): Calcula la desviación estándar y la varianza, respectivamente.
print(empleados.std())
print(empleados.var())
print("-" * 40)

# describe(): Proporciona estadísticas descriptivas, como la media, la desviación
# estándar, el mínimo, el percentil 25, la mediana (50%), el percentil 75 y el máximo.
print(empleados.describe())
print("-" * 40)

# cumsum(): Calcula la suma acumulativa de los elementos.
print(empleados.cumsum())
print("-" * 40)

# cumprod(): Calcula el producto acumulativo de los elementos.
print(empleados.cumprod())
print("-" * 40)

# skew(): Calcula la asimetría de la distribución de datos.
print(empleados.skew())
print("-" * 40)
```

####MÁXIMOS Y MÍNIMOS

```python
# El abanico de posibilidades es muy amplio en cuanto a la búsqueda de valores
# máximos y mínimos en una serie. Veamos lo que nos ofrece pandas a este respecto.
data = {'Apple': 147000,
        'Samsung': 267937,
        'Google': 135301,
        'Microsoft': 163000,
        'Huawei': 197000,
        'Dell': 158000,
        'Facebook': 58604,
        'Foxconn': 878429,
        'Sony': 109700}

empleados = pd.Series(data, name='Cantidad empleados Tech')
print(empleados)
print("-" * 40)

# min() / max(): Encuentra los valores mínimo y máximo, respectivamente.
print(empleados.min())
print(empleados.max())
print("-" * 40)

# Posición (índice) del valor mínimo/máximo de una serie.
print(empleados.argmin())
print(empleados.argmax())
print("-" * 40)

# Etiqueta (índice) del valor mínimo/máximo de una serie.
print(empleados.idxmin())
print(empleados.idxmax())
print("-" * 40)

# Obtener los 2 valores menores/mayores de una serie.
print(empleados.nsmallest(2)) # por defecto devuelve 5 valores
print(empleados.nlargest(2)) # por defecto devuelve 5 valores
print("-" * 40)
```

```python
# Existen múltiples formas de crear un DataFrame en pandas. Veamos algunas de ellas.

# DATAFRAME DESDE DICCIONARIO DE LISTAS
# Cada elemento del diccionario se convierte en una columna, donde su clave es
# el nombre y sus valores se despliegan en «vertical»
data = {'A': [1, 2, 3], 'B': [4, 5, 6]}
print(pd.DataFrame(data))
```

```python
# DATAFRAME DESDE LISTA DE DICCIONARIOS
# Cada elemento de la lista se convierte en una fila. Las claves de cada diccionario
# serán los nombres de las columnas y sus valores se despliegan en «horizontal»
data = [{'A': 1, 'B': 2, 'C': 3}, {'A': 4, 'B': 5, 'C': 6}]
print(pd.DataFrame(data))
```

```python
# DATAFRAME DESDE LISTA DE LISTAS
# Cada elemento de la lista se convierte en una fila y sus valores se despliegan
# en «horizontal». Los nombres de las columnas deben pasarse como parámetro opcional
data = [[1, 2], [3, 4], [5, 6]]
print(pd.DataFrame(data, columns=['A', 'B']))
```

```python
# DATAFRAME DESDE ARRAYS
# Para crear un DataFrame a partir de un array de NumPy
df = pd.DataFrame(np.random.randn(4, 3), columns=["a", "b", "c"])

print(df)
```

```python
# Cuando creamos un DataFrame, pandas autocompleta el índice con un valor entero
# autoincremental comenzando desde cero
stats = pd.DataFrame({'A': [75, 99], 'B': [41, 32]})

# Si queremos convertir alguna columna en el índice de la tabla, podemos hacerlo así
stats.set_index('A')  # columna A como índice
```

# Ejemplos de Series

```python
# En este ejemplo, la Serie "Temperatura_Diaria" representa las temperaturas
# diarias en grados Celsius. Se calcula la temperatura media y se identifican los
# días en los que la temperatura supera la media.

import pandas as pd

# Simular datos de temperaturas diarias en grados Celsius
fecha_inicio = pd.to_datetime("2023-01-01")
fecha_fin = pd.to_datetime("2023-01-10")
rangos_fechas = pd.date_range(fecha_inicio, fecha_fin)
temperaturas_diarias = [25.0, 24.5, 23.8, 26.2, 27.5, 22.8, 23.0, 25.5, 28.0, 26.7]

# Crear una Serie con las temperaturas y fechas como índice
serie_temperaturas = pd.Series(temperaturas_diarias, index=rangos_fechas, name="Temperatura_Diaria")

# Imprimir la Serie de temperaturas diarias
print("Temperaturas diarias (°C):")
print(serie_temperaturas)

# Calcular la temperatura media
temperatura_media = serie_temperaturas.mean()

# Imprimir la media
print("\nLa media de temperaturas:")
print(temperatura_media)

# Encontrar días con temperaturas superiores a la media
dias_calurosos = serie_temperaturas[serie_temperaturas > temperatura_media]

# Imprimir días calurosos
print("\nDías con temperaturas superiores a la media:")
print(dias_calurosos)
```

```python
# creamos una Serie de pandas llamada serie_frecuencia_cardiaca que representa la
# frecuencia cardíaca de un paciente durante cada día de la semana. La Serie tiene
# un índice que son los días de la semana y contiene valores aleatorios de frecuencia cardíaca.
# Luego, mostramos la Serie, accedemos a un valor específico (la frecuencia cardíaca
# del martes), y calculamos algunas estadísticas simples, como el promedio, el mínimo y el máximo.

import pandas as pd
import numpy as np

# Crear datos de ejemplo
dias_semana = ['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes', 'Sábado', 'Domingo']

# estos serian los datos que ingresa en medico al sistema
frecuencia_cardiaca = np.random.randint(60, 100, size=len(dias_semana))

# Crear una Serie de pandas con los datos
serie_frecuencia_cardiaca = pd.Series(frecuencia_cardiaca, index=dias_semana, name='Frecuencia Cardíaca')

# Mostrar la Serie
print("Serie de Frecuencia Cardíaca:")
print(serie_frecuencia_cardiaca)

# Acceder a valores específicos
print("\nFrecuencia Cardíaca del Martes:", serie_frecuencia_cardiaca['Martes'])

# Calcular estadísticas simples
print("\nEstadísticas:")
print("Promedio:", serie_frecuencia_cardiaca.mean())
print("Mínimo:", serie_frecuencia_cardiaca.min())
print("Máximo:", serie_frecuencia_cardiaca.max())
```
