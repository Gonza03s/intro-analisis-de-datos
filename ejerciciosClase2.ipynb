# Ejercicio 1 — Ventas de una tienda

Una tienda registró la cantidad de unidades vendidas durante una semana:

```
ventas = np.array([12,18,15,21,17,25,20])
```


Consignas
- Mostrar el array completo.
- Mostrar la cantidad de ventas registradas.
- Mostrar la cantidad de ventas del primer día.
- Mostrar la cantidad de ventas del último día.
- Mostrar las ventas correspondientes a los días 2, 3 y 4.
- Calcular el total de unidades vendidas durante la semana.
- Calcular el promedio diario de ventas.
- Determinar el día con mayor cantidad de ventas.
- Determinar el día con menor cantidad de ventas.
- Mostrar qué días tuvieron ventas superiores al promedio.

````python
import numpy as np

dias = np.array(["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado","Domingo"])
ventas = np.array([12,18,15,21,17,25,20])
posicion_mayor_venta = np.argmax(ventas)
posicion_menor_venta = np.argmin(ventas)
promedio_ventas = np.mean(ventas)
ventas_mayor_promedio_posiciones = ventas > promedio_ventas

print(f"Array completo: {ventas}")

print(f"Cantidad de ventas registradas: ")
for i in range(len(ventas)):
  print(f"{dias[i]} - venta: {ventas[i]}")

print(f"ventas registradas en el primer dia: {ventas[0]}")
print(f"ventas registradas en el ultimo dia: {ventas[-1]}")
print(f"ventas correspondientes a los días 2, 3 y 4 {ventas[1:4]} ")
print(f"unidades vendidas durante la semana: {np.sum(ventas)}")
print(f"promedio diario de ventas: {promedio_ventas:.2f}")
print(f"dia con mayor cantidad de ventas: {dias[posicion_mayor_venta]}" )
print(f"dia con menor cantidad de ventas: {dias[posicion_menor_venta]}" )
print(f"ventas superiores al promedio:\nDia:{dias[ventas_mayor_promedio_posiciones]} - cantidad: {ventas[ventas_mayor_promedio_posiciones]} ")



````

consignas:
- Calcular el salario promedio.
- Determinar el salario mínimo.
- Determinar el salario máximo.
- Calcular la diferencia entre el salario máximo y mínimo.

- Mostrar los salarios superiores al promedio.
- Determinar cuántos empleados cobran más de $1.300.000.

- Aplicar un aumento del 15% a todos los salarios.
- Calcular el nuevo salario promedio.
- Determinar cuánto aumentó el salario promedio.


````python
import numpy as np
salarios = np.array([
    850000, 920000, 1100000, 1250000,
    980000, 1450000, 1600000, 870000,
    1320000, 1180000, 1750000, 990000
])

promedio_original = np.mean(salarios)
diferencia = np.max(salarios) - np.min(salarios)

print(f"salario promedio: {promedio_original:,.2f}")
print(f"salario minimo: {np.min(salarios):,}")
print(f"salario maximo: {np.max(salarios):,}")
print(f"diferencia entre el salario máximo y mínimo.: {diferencia:,}")
print(f"cantidad empleados que cobran mas que el salario promedio: {np.sum([salarios > promedio_original])}")
print(f"salarios superiores al promedio: {salarios[salarios > promedio_original]}")
print(f"cantidad de empleados que cobran mas de $1.300.000: {np.sum(salarios > 1300000)}")
print(f"salarios mayores a $1.300.000: {salarios[salarios > 1300000]}")

salarios = salarios * 1.15
nuevo_promedio = np.mean(salarios)
diferencia_promedios = nuevo_promedio - promedio_original
diferencia_porcentual = (diferencia_promedios / promedio_original) * 100


print(f"salarios promedio con 15% de aumento: {nuevo_promedio:,.2f}")
print(f"el salario promedio aumento un: {diferencia_porcentual:.2f}%")
print(f"el salario promedio aumento un: {diferencia_promedios:.2f}")
````

# Ejercicio 5 — Rendimiento de estudiantes
Una institución registró las calificaciones obtenidas por 20 estudiantes:
```
notas = np.array([
    6, 8, 7, 9, 4,
    10, 5, 7, 8, 6,
    3, 9, 8, 7, 10,
    6, 5, 9, 4, 8
])
```
Consignas
- Calcular el promedio general.
- Determinar la nota máxima.
- Determinar la nota mínima.
- Determinar cuántos estudiantes aprobaron.
- Considerar aprobado una nota mayor o igual a 6.
- Mostrar las notas de los estudiantes aprobados.
- Mostrar las notas de los estudiantes desaprobados.
- Determinar el porcentaje de estudiantes aprobados.
- Determinar el porcentaje de estudiantes desaprobados.
- Determinar cuántos estudiantes obtuvieron una nota superior al promedio

¿Qué diferencia existe entre:
- notas >= 6
- notas[notas >= 6]

- notas >= 6: se obtiene un vector con valores booleanos los cuales cumplen con esa condicion

- notas[notas >= 6]: al anteponer el nombre de la variable y dentro de los corchetes poner la condicion, accedes a los valores que cumplen dicha condicion

````python

import numpy as np

notas = np.array([
    6, 8, 7, 9, 4,
    10, 5, 7, 8, 6,
    3, 9, 8, 7, 10,
    6, 5, 9, 4, 8
])

promedio = np.mean(notas)
total_estudiantes = len(notas)
estudiantes_aprobados = np.sum(notas >= 6)
estudiantes_desaprobados = np.sum(notas < 6)
porcentaje_aprobados = (estudiantes_aprobados * 100) / total_estudiantes
porcentaje_desaprobados = (estudiantes_desaprobados * 100) / total_estudiantes
notas_superiores_promedio = notas[notas > promedio]

print(f"Promedio de notas: {promedio:.2f}")
print(f"nota maxima: {np.max(notas)}")
print(f"nota minima: {np.min(notas)}")
print(f"cantidad estudiantes aprobados: {estudiantes_aprobados}")
print(f"notas de estudiantes aprobados: {notas[notas >= 6]}")
print(f"notas de estudiantes desaprobados: {notas[notas < 6]}")
print(f"porcentaje de estudiantes aprobados: {porcentaje_aprobados:.0f}%")
print(f"porcentaje de estudiantes desaprobados: {porcentaje_desaprobados:.0f}%")
print(f"Cantidad de estudiantes con notas mayor al promedio: {len(notas_superiores_promedio)}")
````

# Ejercicio 8 — Broadcasting aplicado a ventas
- Una empresa vende cuatro productos
```
 precios = np.array([12000, 18500, 9500, 22000])
```
- Durante tres meses se vendieron:
```
 ventas = np.array([
    [10, 15, 12, 8],
    [14, 18, 15, 11],
    [20, 22, 18, 15]
])
```
Consignas
- Mostrar las dimensiones de ventas.
- Determinar cuántos meses y productos contiene.
- Calcular las unidades vendidas de cada producto durante los tres meses.
- Calcular las unidades vendidas de cada mes.
- Calcular la facturación de cada producto para cada mes.
- Calcular la facturación total de cada mes.
- Calcular la facturación total de cada producto.
- Determinar qué producto generó mayor facturación.
- Determinar qué mes tuvo mayor facturación.
- Aumentar todos los precios un 10%.
- Volver a calcular la facturación utilizando los nuevos precios.

¿Por qué podemos hacer ventas * precios si ventas tiene dos dimensiones y precios solamente una?


````python
import numpy as np

precios = np.array([12000, 18500, 9500, 22000])

ventas = np.array([
    [10, 15, 12, 8],
    [14, 18, 15, 11],
    [20, 22, 18, 15]
])
#broadcasting cruza la columna con el precio correspondiente

facturacion = ventas * precios

print(f"Dimensiones de ventas: {np.ndim(ventas)}")
print(f"Meses: {np.shape(ventas)[0]} - Productos: {np.size(precios)}")
print(f"unidades vendidas de cada producto durante los tres meses: {np.sum(ventas,axis=0)}")
print(f"unidades vendidas de cada mes: {np.sum(ventas,axis=1)}")
print(f"\nfacturacion de cada producto para cada mes: \n{facturacion}")
print(f"\nfacturacion total de cada mes:{np.sum(facturacion,axis=1)}")
print(f"facturacion total de cada producto: {np.sum(facturacion,axis=0)}")

facturacion_por_producto = np.sum(facturacion, axis=0)
posicion_producto_mas_vendido = np.argmax(facturacion_por_producto)
valor_producto_mas_vendido= facturacion_por_producto[posicion_producto_mas_vendido]

print(f"producto que genero la mayor facturacion fue: {posicion_producto_mas_vendido + 1} con: {valor_producto_mas_vendido}")

facturacion_mensual = np.sum(facturacion,axis=1)
posicion_mes_mayor_facturado = np.argmax(facturacion_mensual)

print(f"el mes que tuvo mayor facturacion fue: {posicion_mes_mayor_facturado +1} con: {facturacion_mensual[posicion_mes_mayor_facturado]}")

precios = precios * 1.10
facturacion = ventas * precios
print(f"\nFacturacion con un 10% de aumento a los precios:\n{facturacion}")
````

```
ventas = np.array([
```

consignas

- Calcular el promedio de ventas de cada mes.

# Parte 10 — Simulación de aumento de precios
La empresa está evaluando aumentar los precios un 8%.
Crear un nuevo array con los precios actualizados.
Luego:

- Mostrar los precios originales.
- Mostrar los nuevos precios.
- Calcular nuevamente la facturación.
- Comparar la facturación original con la facturación proyectada.
- Determinar cuánto aumentaría la facturación total si las cantidades vendidas permanecieran constantes.

Pregunta
- ¿Qué supuesto estamos haciendo al realizar esta simulación?

- el supuesto que estamos realizando concretamente es la inelasticidad de la demanda. el mismo implica en este caso que a pesar del aumento de precio de los productos
las cantidad vendidas no bajaran a pesar del aument del 8%

````python
precios_con_aumento = precios * 1.08
facturacion_con_aumento = ventas * precios_con_aumento

facturacion_total = np.sum(facturacion_total_cada_producto)
facturacion_total_proyectada= np.sum(facturacion_con_aumento)

diferencia = facturacion_total_proyectada - facturacion_total

print("Comparativa de precios:")

print(f"Precios originales: {precios}")
print(f"Precios con aumento de 8%: {precios_con_aumento}")

print("\nComparativa de facturacion:")

print(f"Facturacion con aumento de 8%: {np.sum(facturacion_con_aumento):,.0f}")
print(f"Facturacion original: {facturacion_total:,.0f} - facturacion proyectada: {facturacion_total_proyectada:,.0f}")
print(f"aumentaria: ${diferencia:,.0f}")

````

# Parte 14 — Crear un indicador
La empresa quiere clasificar los meses según su nivel de ventas.
Utilizar el total de unidades vendidas por mes.
Definir:
- Menos de 180 unidades → Bajo
- Entre 180 y 220 → Medio
- Más de 220 → Alto

Determinar:

- Qué categoría corresponde a cada mes.
- Cuántos meses pertenecen a cada categoría.
- Qué porcentaje representa cada categoría.


````python

categorias_meses = []

for unidades in unidades_vendidas_mensual:
    if unidades < 180:
        categorias_meses.append("Bajo")
    elif unidades >= 180 and unidades <= 220:
        categorias_meses.append("Medio")
    else:
        categorias_meses.append("Alto")

categorias_meses = np.array(categorias_meses)

print("clasificacion mensual: ")
for mes, cat, unids in zip(meses, categorias_meses, unidades_vendidas_mensual):
    print(f"{mes}: {cat} ({unids} unidades)")

total_meses = len(meses)
print("\nresumen por categoria:")
for cat in ["Bajo", "Medio", "Alto"]:
    cantidad = np.sum(categorias_meses == cat)
    porcentaje = (cantidad / total_meses) * 100
    print(f"Categoría {cat}: {cantidad} meses ({porcentaje:.1f}%)")

````
