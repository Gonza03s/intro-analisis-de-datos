**¿Qué característica tienen en común los datos estructurados?**

- Los datos estructurados, tienen la caracteristica en comun que siguen un esquema tabular de filas y columnas

**¿Por qué JSON y XML no se consideran datos completamente estructurados?**

- Los mismos refieren a datos semi-estructurados, esto es porque no siguen el enfoque tabular SQL de filas y columnas, en cambio presentan etiquetas, claves que permiten la identificacion de informacion

**¿Qué diferencia principal existe entre un dato semiestructurado y uno no estructurado?**

- La principal diferencia es que los datos semiestructurados poseen una organizacion interna basada en etiquetas o relaciones clave-valor (ej. JSON), lo que facilita su lectura. Por el contrario, los datos no estructurados no presentan ninguna estructura de filas, columnas o etiquetas, y corresponden tipicamente a archivos de audio, video, imágenes o texto libre.

**¿Un archivo Excel siempre contiene datos estructurados? Justificá.**

- No siempre. Aunque Excel utiliza una estructura visual de filas y columnas, la clasificación depende de la organizacion del contenido. Un Excel contiene datos estructurados solo cuando sigue un esquema estricto (una unica tabla limpia, con encabezados claros y tipos de datos uniformes por columna). Si la planilla incluye celdas combinadas, multiples tablas dispersas, tipos de datos inconsistentes, texto libre o imagenes, se considera un dato semiestructurado o incluso una mezcla con elementos no estructurados.

# Parte 1 — Crear el dataset

```python
import random
random.seed(42)

nombres = [
"Ana", "Bruno", "Carla", "Diego", "Elena",
"Federico", "Gabriela", "Hugo", "Irene", "Javier",
"Karen", "Lucas", "Mariana", "Nicolas", "Olivia",
"Pablo", "Romina", "Santiago", "Tamara", "Ulises",
"Valentina", "Walter", "Ximena", "Yamila", "Zoe"
]

apellidos = [
"Gomez", "Rodriguez", "Fernandez", "Lopez", "Martinez",
"Perez", "Garcia", "Sanchez", "Romero", "Torres",
"Diaz", "Alvarez", "Ruiz", "Sosa", "Acosta",
"Benitez", "Medina", "Herrera", "Suarez", "Rojas"
]

departamentos = [
"IT",
"Ventas",
"Marketing",
"RRHH",
"Finanzas",
"Soporte"
]

empleados = []

for i in range(1, 239):
  nombre = random.choice(nombres)
  apellido = random.choice(apellidos)

  empleado = {
  "id": i,
  "nombre": f"{nombre} {apellido}",
  "edad": random.randint(22, 60),
  "departamento": random.choice(departamentos),
  "salario": random.randint(650000, 1800000),
  "experiencia": random.randint(0, 25)
  }
  empleados.append(empleado)
```

# 3.4 Valores extremos
Determinar:
- La edad mínima.
- La edad máxima.
- El salario mínimo.
- El salario máximo.
- La menor cantidad de años de experiencia.
- La mayor cantidad de años de experiencia.

```python
def obtener_extremos(lista, clave):
    valores = [item[clave] for item in lista]
    return min(valores), max(valores)

edad_minima, edad_maxima = obtener_extremos(empleados, "edad")
salario_minimo, salario_maximo = obtener_extremos(empleados, "salario")
exp_minima, exp_maxima = obtener_extremos(empleados, "experiencia")


print(f"La edad minima es: {edad_minima}")
print(f"La edad maxima es: {edad_maxima}")

print(f"el salario minimo es: {salario_minimo:,.0f}")
print(f"el salario maximo es: {salario_maximo:,.0f}")

print(f"la experiencia minima es: {exp_minima}")
print(f"la experiencia maxima es: {exp_maxima}")
```

#Parte 7 — Crear una nueva variable
La empresa quiere clasificar a sus empleados según su experiencia profesional.
Utilizar las siguientes categorías:

| Experiencia  | Nivel |
| --- | --- |
| Menos de 3 años  | Junior |
| Entre 3 y 7 años | Semi-Senior |
| 8 años o más  | Senior |

Agregar a cada empleado una nueva clave llamada:
"nivel"

Luego:
- Contar cuántos empleados hay en cada nivel.
- Mostrar los empleados clasificados como Senior.
- Determinar qué porcentaje de los empleados pertenece a cada nivel.


```python

niveles = {"Junior": 0, "Semi-Senior": 0, "Senior": 0}
total_empleados = len(empleados)

for emp in empleados:
  if emp["experiencia"] < 3:
      emp["nivel"] = "Junior"
  elif  emp["experiencia"] >= 3 and emp["experiencia"] <= 7:
      emp["nivel"] = "Semi-Senior"
  elif emp["experiencia"] >= 8:
      emp["nivel"] = "Senior"

  niveles[emp["nivel"]] +=1


for nivel in niveles:
    print(f" {nivel} - {niveles[nivel]} ")

empleados_seniors = [emp for emp in empleados if emp["nivel"] == "Senior"]
print(f"\nEmpleados seniors: {empleados_seniors}\n")

for nivel in niveles:
  print(f"porcentaje {nivel}: {niveles[nivel] * 100 / total_empleados:.2f}")
```

# 8.2 Salario promedio por departamento
Calcular el salario promedio de cada departamento.
Por ejemplo:
- IT → $

- Ventas → $

- Marketing → $

- RRHH → $

- Finanzas → $

- Soporte → $


```python
suma_salarios ={}
conteo_deptos={}

for emp in empleados:
  depto= emp["departamento"]
  salario= emp["salario"]

  suma_salarios[depto] = suma_salarios.get(depto,0) + salario
  conteo_deptos[depto] = conteo_deptos.get(depto,0) + 1

for depto in suma_salarios:
  promedio = suma_salarios[depto] / conteo_deptos[depto]
  print(f"{depto}: {promedio:.2f}")
```

# 8.3 Departamento con mayor salario promedio
Determinar qué departamento tiene el salario promedio más alto

```python
suma_salarios ={}
conteo_deptos={}
salario_promedio_deptos = {}
bandera = 0
salario_mayor = 0
depto_maximo = ""

for emp in empleados:
  depto= emp["departamento"]
  salario= emp["salario"]

  suma_salarios[depto] = suma_salarios.get(depto,0) + salario
  conteo_deptos[depto] = conteo_deptos.get(depto,0) + 1

for depto in suma_salarios:
  promedio = suma_salarios[depto] / conteo_deptos[depto]
  salario_promedio_deptos[depto] = promedio


for depto,promedio in salario_promedio_deptos.items():
  if bandera == 0 or promedio > salario_mayor:
    bandera=1
    salario_mayor = promedio
    depto_maximo= depto

print(f"departamento con mayor salario promedio:{depto_maximo}\ncuenta con: {salario_mayor:,.0f}")


```

# 8.4 Departamento con mayor cantidad de empleados

```python
contador_departamentos = {}
bandera=0
cantidad_emp_mayor = 0
depto_mayor = ""

for emp in empleados:
  depto = emp["departamento"]
  contador_departamentos[depto] = contador_departamentos.get(depto, 0) + 1

for depto,cantidad_emp in contador_departamentos.items():
  if bandera == 0 or cantidad_emp > cantidad_emp_mayor:
    bandera = 1
    cantidad_emp_mayor = cantidad_emp
    depto_mayor = depto

print(f"Departamento con mayor cantidad de empleados: {depto_mayor} empleados {cantidad_emp_mayor}")

# depto_mayor = max(contador_departamentos, key=contador_departamentos.get)
# cantidad_emp_mayor = contador_departamentos[depto_mayor]
```

```python
def obtener_edad_promedio(lista:list)->float:
  acc_edades= 0
  cantidad_items = len(lista)

  for item in lista:
    acc_edades += item["edad"]

  return acc_edades / cantidad_items

def obtener_salario_promedio(lista:list)->float:
  acc_salario = 0
  cantidad_items = len(lista)

  for item in lista:
    acc_salario += item["salario"]

  return acc_salario / cantidad_items

def obtener_experiencia_promedio(lista:list)->int:
  acc_experiencia = 0
  cantidad_items = len(lista)

  for item in lista:
    acc_experiencia += item["experiencia"]

  return acc_experiencia / cantidad_items

def obtener_mayor_salario(lista:list)->dict:
  mayor_salario = 0
  bandera = 0
  item_mayor_salario = {}

  for item in lista:
    if bandera == 0 or item["salario"] > mayor_salario:
      mayor_salario = item["salario"]
      item_mayor_salario = item
      bandera = 1

  return item

def obtener_departamento_mayor_salario(lista: list) -> list:
    suma_salarios = {}
    conteo_deptos = {}
    salario_promedio_deptos = {}
    bandera = 0
    salario_mayor = 0
    depto_maximo = ""

    for emp in lista:
        depto = emp["departamento"]
        salario = emp["salario"]

        suma_salarios[depto] = suma_salarios.get(depto, 0) + salario
        conteo_deptos[depto] = conteo_deptos.get(depto, 0) + 1

    for depto in suma_salarios:
        promedio = suma_salarios[depto] / conteo_deptos[depto]
        salario_promedio_deptos[depto] = promedio

    for depto, promedio in salario_promedio_deptos.items():
        if bandera == 0 or promedio > salario_mayor:
            bandera = 1
            salario_mayor = promedio
            depto_maximo = depto

    salario_formateado = f"{salario_mayor:,.2f}"

    return [depto_maximo, salario_formateado]


def obtener_departamento_mayor_empleados(lista:list)->list:
  contador_departamentos = {}
  cantidad_emp_mayor = 0
  depto_mayor = ""

  for emp in lista:
    depto = emp["departamento"]
    contador_departamentos[depto] = contador_departamentos.get(depto, 0) + 1

  depto_mayor = max(contador_departamentos, key=contador_departamentos.get)
  cantidad_emp_mayor = contador_departamentos[depto_mayor]

  return [depto_mayor,cantidad_emp_mayor]

def contar_empleados_por_nivel(lista:list)->dict:
  conteo_niveles = {}

  for item in lista:
      nivel= item["nivel"]
      conteo_niveles[nivel] = conteo_niveles.get(nivel,0)+1

  return conteo_niveles


print(f"\ncantidad total emps: {len(empleados)}")
print(f"\nedad promedio: {obtener_edad_promedio(empleados):.2f}")
print(f"\nsalario promedio {obtener_salario_promedio(empleados):,.0f}")
print(f"\nexperiencia promedio: {obtener_experiencia_promedio(empleados):.2f}")
print(f"\nemp mayor salario: {obtener_mayor_salario(empleados)}")
print(f"\ndepto mayor salario: {obtener_departamento_mayor_salario(empleados)}")
print(f"\ndepto con mayor cantidad emps: {obtener_departamento_mayor_empleados(empleados)}")
print(f"\ncantidad empleados por nivel: {contar_empleados_por_nivel(empleados)}")
```
